```python
import matplotlib.pyplot as plt   # plotting library
import numpy as np                # scientific computing library

rng = np.random.default_rng(42)
x = np.linspace(0, 10, 300)

dragons = 60 + 40 * np.sin(x) + 8 * rng.normal(size=x.size)
gold = 20 + 6 * x + 15 * np.sin(1.3 * x + 1)
villages = 10 + 3 * x ** 1.2
legend_pts = 5 + 9 * np.exp(0.15 * x) * np.abs(np.sin(0.5 * x))

colors = {
    "Dragons Sighted": "#2a78d6",     # blue
    "Gold Hoarded (kg)": "#eb6834",   # orange
    "Villages Saved": "#1baf7a",      # aqua
    "Legend Points": "#eda100",       # yellow
}
data = {
    "Dragons Sighted": dragons,
    "Gold Hoarded (kg)": gold,
    "Villages Saved": villages,
    "Legend Points": legend_pts,
}

fig, ax = plt.subplots(figsize=(11, 6.5))
fig.patch.set_facecolor("#fcfcfb")
ax.set_facecolor("#fcfcfb")

for label, y in data.items():
    ax.plot(x, y, label=label, color=colors[label], linewidth=3, solid_capstyle="round")

# Annotate the peak of the fastest-climbing series
peak_idx = np.argmax(legend_pts)
ax.annotate(
    "PEAK LEGEND STATUS",
    xy=(x[peak_idx], legend_pts[peak_idx]),
    xytext=(x[peak_idx] - 3.2, legend_pts[peak_idx] + 18),
    arrowprops=dict(arrowstyle="->", color=colors["Legend Points"], linewidth=2),
    fontsize=12, fontweight="bold", color="#0b0b0b",
)

# Call out where the steady climb begins
climb_idx = np.argmin(np.abs(x - 2))
ax.annotate(
    "Steady climb begins",
    xy=(x[climb_idx], villages[climb_idx]),
    xytext=(0.3, 55),
    arrowprops=dict(arrowstyle="->", color=colors["Villages Saved"], linewidth=2),
    fontsize=10, fontweight="bold", color="#0b0b0b",
)

ax.set_title("THE EPIC RISE OF THE DRAGON SLAYER", fontsize=19, fontweight="bold", color="#0b0b0b", pad=16)
ax.set_xlabel("Chapters of the Saga", fontsize=12, color="#52514e")
ax.set_ylabel("Legendary Units", fontsize=12, color="#52514e")

ax.grid(True, linestyle="--", linewidth=0.7, color="#e1e0d9")
ax.set_axisbelow(True)

for spine in ("top", "right"):
    ax.spines[spine].set_visible(False)
for spine in ("left", "bottom"):
    ax.spines[spine].set_color("#c3c2b7")

ax.legend(loc="upper left", frameon=True, fontsize=10, facecolor="#fcfcfb", edgecolor="#c3c2b7")

plt.tight_layout()
plt.show()
```
